```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Import Data

titanicData = pd.read_excel('Titanic.xlsx')
titanicData.head()
```

# Descriptive Analysis

```python

# Dataset Columns and Rows
print("Titanic Dataset Column and Rows: {}".format(titanicData.shape))
print()

# Dataset Column Names
print("Titanic Dataset Column names:")
print(titanicData.columns)
print()

# Soal Nomor 1
# Basic Data Statistics
print("Titanic Dataset Basic Statistical Data: ")
titanicData.describe().T
# print()
```

### More Detailed Statistics

#### Survived Statistics

```python
# Average Survival Chance
averageSurvivalChance = titanicData['Survived'].mean() * 100
print("Titanic Average Survival Chance: %.0f%%" % averageSurvivalChance)

```

##### By Sex

```python
# By Sex
# Male
maleAverageSurvChance = titanicData[titanicData["Sex"] == "male"]['Survived'].mean() * 100
print("Male Average Survival Chance: %.0f%%" % maleAverageSurvChance)

# Female
# Soal Nomor 2
femaleAverageSurvChance = titanicData[titanicData["Sex"] == "female"]['Survived'].mean() * 100
print("Female Survival Chance: %.0f%%" % femaleAverageSurvChance)

# Soal Nomor 4
firstClassFemaleSurvChance = titanicData[(titanicData["Sex"] == "female") & (titanicData['Pclass'] == 1)]['Survived'].mean() * 100
print("First Class Female Survival Chance: %.0f%%" % firstClassFemaleSurvChance)
```

##### By Class

```python
# By Class
# Soal Nomor 3

classThreeSurvChance = titanicData[titanicData['Pclass'] == 3]['Survived'].mean() * 100
print("Third Class Passanger Survival Rate: %0.f%%" % classThreeSurvChance)

classTwoSurvChance = titanicData[titanicData['Pclass'] == 2]['Survived'].mean() * 100
print("Second Class Passanger Survival Rate: %0.f%%" % classTwoSurvChance)

classOneSurvChance = titanicData[titanicData['Pclass'] == 1]['Survived'].mean() * 100
print("First Class Passanger Survival Rate: %0.f%%" % classOneSurvChance)

print("\nSurvival Rate by Passanger Class Bar Distribution")
pClass = ["First Class", "Second Class", "Third Class"]
survRateByClass = [classOneSurvChance, classTwoSurvChance, classThreeSurvChance]

plt.bar(pClass, survRateByClass, color='blue')
plt.title("Rate of Survival by Passanger Class")
plt.ylabel("Rate of Survival (%)")
plt.ylim(0,100)

plt.show()
```

##### By Age

```python
# By Age
youngSurvChance = titanicData[titanicData['Age'] < 20]['Survived'].mean() * 100
print("Average Chance Someone under 20 Survived: %.0f%%" % youngSurvChance)

adultSurvChance = titanicData[titanicData['Age'].between(20, 50)]['Survived'].mean() * 100
print("Average Chance Someone in the range of 20 and 50 Survived: %.0f%%" % adultSurvChance)

oldSurvChance = titanicData[titanicData['Age'] > 50]['Survived'].mean() * 100
print("Average Chance Someone older than 50 Survived: %.0f%%" % oldSurvChance)

# Soal Nomor 5
# Survival Distribution by Age
survivedAgeDistribution = list(titanicData[titanicData['Survived'] == 1]['Age'])
notSurvivedAgeDistribution = list(titanicData[titanicData['Survived'] == 0]['Age'])

plt.hist([survivedAgeDistribution, notSurvivedAgeDistribution], label=['Survived', 'Not Survived'], bins=10)

plt.title("Survived and Not Survived Age Distribution")
plt.xlabel('Age')
plt.ylabel('Passangers')
plt.ylim(0,140)
plt.xlim(0,90)
plt.legend()

plt.show()
```
