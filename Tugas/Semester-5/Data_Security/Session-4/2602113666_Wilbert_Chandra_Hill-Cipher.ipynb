```python
import numpy as np

def hill_cipher_encrypt(message, key):
    # Remove spaces and convert to uppercase
    message = message.replace(" ", "").upper()
    
    # Pad the message if necessary
    block_size = int(np.sqrt(len(key)))
    padding_length = (block_size - (len(message) % block_size)) % block_size
    message += 'X' * padding_length
    
    # Convert message to numbers (A=0, B=1, ..., Z=25)
    message_nums = [ord(char) - ord('A') for char in message]
    
    # Reshape the key into a matrix
    key_matrix = np.array(key).reshape((block_size, block_size))
    
    # Encrypt the message
    ciphertext = ""
    for i in range(0, len(message_nums), block_size):
        block = np.array(message_nums[i:i+block_size])
        encrypted_block = np.dot(key_matrix, block) % 26
        ciphertext += ''.join([chr(num + ord('A')) for num in encrypted_block])
    
    return ciphertext, padding_length
```

```python
def hill_cipher_decrypt(ciphertext, key, padding_length=0):
    # Remove spaces and convert to uppercase
    ciphertext = ciphertext.replace(" ", "").upper()
    
    # Convert ciphertext to numbers (A=0, B=1, ..., Z=25)
    ciphertext_nums = [ord(char) - ord('A') for char in ciphertext]
    
    # Reshape the key into a matrix
    block_size = int(np.sqrt(len(key)))
    key_matrix = np.array(key).reshape((block_size, block_size))
    
    # Calculate the modular multiplicative inverse of the key matrix
    det = int(np.round(np.linalg.det(key_matrix)))
    det_inv = pow(det, -1, 26)
    adj_matrix = np.round(det * np.linalg.inv(key_matrix)).astype(int)
    inv_key_matrix = (det_inv * adj_matrix % 26).astype(int)
    
    # Decrypt the message
    plaintext = ""
    for i in range(0, len(ciphertext_nums), block_size):
        block = np.array(ciphertext_nums[i:i+block_size])
        decrypted_block = np.dot(inv_key_matrix, block) % 26
        plaintext += ''.join([chr(num + ord('A')) for num in decrypted_block])
    
    # Remove padding
    if padding_length > 0:
        plaintext = plaintext[:-padding_length]
    
    return plaintext
```

```python

# Example usage
key = [2, 1, 3, 4]  # key matrix
plaintext = "KRIPTO"
encrypted, padding_length = hill_cipher_encrypt(plaintext, key)
decrypted = hill_cipher_decrypt(encrypted, key, padding_length)

print(f"Plaintext: {plaintext}")
print(f"Encrypted: {encrypted}")
print(f"Decrypted: {decrypted}")
```
