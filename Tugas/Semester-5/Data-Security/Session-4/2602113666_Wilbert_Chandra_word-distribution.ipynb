```python
import requests
from bs4 import BeautifulSoup
from collections import Counter
import matplotlib.pyplot as plt
import re

def fetch_webpage(url):
    response = requests.get(url)
    return response.text

def extract_text(html):
    soup = BeautifulSoup(html, 'html.parser')
    return soup.get_text()

def process_text(text):
    # Convert to lowercase and keep only alphabet characters
    return re.findall(r'[a-z]', text.lower())

def calculate_char_distribution(chars):
    return Counter(chars)

def visualize_distribution(char_dist):
    # Sort the distribution by keys (a to z)
    sorted_dist = dict(sorted(char_dist.items()))
    
    plt.figure(figsize=(12, 6))
    plt.bar(sorted_dist.keys(), sorted_dist.values())
    plt.title('Alphabet Character Distribution')
    plt.xlabel('Characters')
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

# 1. Fetch webpage
url = "https://example.com" 
html = fetch_webpage(url)

# 2. Extract and process text
text = extract_text(html)
chars = process_text(text)

# 3. Calculate distribution
char_dist = calculate_char_distribution(chars)

# 4. Visualize
visualize_distribution(char_dist)

# 5. Display top 5 most common characters
print("Top 5 most common characters:")
for char, count in char_dist.most_common(5):
    print(f"{char}: {count}")
```
