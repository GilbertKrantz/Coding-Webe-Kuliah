```python
from PIL import Image, ImageSequence
import numpy as np
#open tiff image 
im = Image.open("train2/masks/tiff/115_jpg.rf.7d59e0053397d75d0fdf2af67bbaa8f0_seg_0.tif")

#navigate to the folder were the layers are going to be saved 


#loop over layers and export jpeg instances
for i, page in enumerate(ImageSequence.Iterator(im)):
    page = page.convert('L')
    page = page.point(lambda i:i*(1./256))
    page = page.convert('L')
    page.save(str(i)+'.jpeg')
```
