# Change Columns Name to short name

```python
data['Nama Lengkap'] = data['Nama Lengkap Contoh : Angelina Jolie ']
data.drop(columns=['Nama Lengkap Contoh : Angelina Jolie '], axis=1, inplace=True)

data['NIM'] = data['NIM Contoh : 2602728913 '] 
data.drop(columns=['NIM Contoh : 2602728913 '], axis=1, inplace=True)
```

```python
data['ITDS'] = data['Berapa nilai akhir Introduction to Data Science yang kamu peroleh?']
data.drop(columns=['Berapa nilai akhir Introduction to Data Science yang kamu peroleh?'], axis=1, inplace=True)
```

```python
data['DMV'] = data['Berapa nilai UTS Data Mining and Visualization yang kamu peroleh?']
data.drop(columns=['Berapa nilai UTS Data Mining and Visualization yang kamu peroleh?'], axis=1, inplace=True)
```

```python
string = 'Question'

for i, column in enumerate(data.columns):
    if column != 'Nama Lengkap' and column != 'NIM' and column != 'Total Score' and column != 'ITDS' and column != 'DMV':
        name = string + str(i+1)
        data[name] = data[column]
        data.drop(columns=[column], axis=1, inplace=True)
```

# Extract Features

```python
# Extract Year from NIM first 2 digits
data['Year'] = data['NIM'].astype(str).str[:2]
```

```python
# Make year the 3rd column
cols = data.columns.tolist()
cols = cols[:1] + cols[-1:] + cols[1:-1]
data = data[cols]
```

# Preprocessing

We will group each question to it's respected measure

### Missing Values Impuation

```python
for i, column in enumerate(data.columns):
    if column != 'Nama Lengkap' and column != 'NIM' and column != 'Total Score' and column != 'ITDS' and column != 'DMV':
        data[column] = data[column].fillna(data[column].mode().values[0])
```

## Distribution of Features

## Correlation

### Whole Dataset

```python
fig = plt.figure(figsize=(20, 10))

sns.heatmap(data.drop(columns=['Nama Lengkap', 'NIM']).corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
```

### Correlation to Year

```python
data_numeric = data.drop(columns=['Nama Lengkap', 'NIM'])

year_mean = data_numeric.groupby('Year').mean().reset_index()

fig = plt.figure(figsize=(20, 20))

for i, column in enumerate(data.columns):
    if column != 'Nama Lengkap' and column != 'NIM' and column != 'Year' and column != 'ITDS' and column != 'DMV' and column != 'Total Score':
        plt.subplot(3, 3, i-5)
        sns.barplot(x='Year', y=column, data=year_mean)
        plt.title(column)
        plt.xlabel(column)
    
plt.tight_layout()
plt.show()
```

On average all year can have the same capabilities in answering the questions. WIth year 27 leading in understanding Basic Statistic by theory and Data Understanding by Image, year 26 leading in Understanding data by theory, and year 25 leading in Data Analysis Application.

```python
fig = plt.figure(figsize=(20, 10))

plt.subplot(1, 3, 1)

sns.barplot(x='Year', y='ITDS', data=year_mean)
plt.title('ITDS')
plt.xlabel('Year')

plt.subplot(1, 3, 2)
sns.barplot(x='Year', y='DMV', data=year_mean)
plt.title('DMV')
plt.xlabel('Year')

plt.subplot(1, 3, 3)
sns.barplot(x='Year', y='Total Score', data=year_mean)
plt.title('Total Score')
plt.xlabel('Year')

plt.tight_layout()
plt.show()
```

On average on ITDS and Total Score score of all year is the same. But the students from year 25 have the highest average score on ITDS and year 27 have the lowest average score on Total Score. For the DMV the students from year 25 also have the highest average score and year 27 have the lowest average score with year 26 being in the middle.
