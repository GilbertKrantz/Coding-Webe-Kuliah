```python
import matplotlib.pyplot as plt
import numpy as np
```

Number 1

```python
arr1 = [72, 74, 76, 77, 81, 84, 85, 91]
print(np.mean(arr1))
print(np.var(arr1))
```

C

```python
for_hist = []
a = 1
for i in arr1:
    for j in arr1:
        if i == j:
            continue
        if [i, j] not in for_hist and [j, i] not in for_hist:
            print(f'[{i}, {j}]: y={(i+j)/2}')
            for_hist.append([i, j])
            a += 1
print(a)
```

```python
sampling_mean = []
for i in for_hist:
    sampling_mean.append((i[0] + i[1]) / 2)
```

```python
plt.hist(sampling_mean)
```

D

```python
pop_avg = np.mean(arr1)

# Sampling variance
print(sum([(i - pop_avg)**2 for i in sampling_mean]))

sampling_variance = sum([(i - pop_avg)**2 for i in sampling_mean]) / len(sampling_mean)
print(f' = {sampling_variance}')
```

```python
import itertools
sampling_variance_6 = []
for i in itertools.combinations(arr1, 6):
    print(f'{i}: y={np.round(np.mean(i),3)}')
    sampling_variance_6.append(np.round(np.mean(i),3))
```

```python
plt.hist(sampling_variance_6)
```

```python
low_scores = [72, 74, 76, 77]
high_scores = [81, 84, 85, 91]

low_scores_mean = np.mean(low_scores)
high_scores_mean = np.mean(high_scores)

print(f'low_scores_mean: {low_scores_mean}')
print(f'high_scores_mean: {high_scores_mean}')
```

```python
low_scores_variance = (sum([(i - low_scores_mean)**2 for i in low_scores]) / 3)
high_scores_variance = (sum([(i - high_scores_mean)**2 for i in high_scores]) / 3)

print(f'low_scores_variance: {low_scores_variance}')
print(f'high_scores_variance: {high_scores_variance}')
```

```python
import itertools
scores = []
sample_2 = []
sampling_mean_2 = []
for i in itertools.combinations(low_scores, 3):
    print(f'{i}: y={np.round(np.mean(i),3)}')
    scores.append(i)

for i in itertools.combinations(high_scores, 3):
    print(f'{i}: y={np.round(np.mean(i),3)}')
    scores.append(i)

for i in itertools.combinations(scores, 2):
    print(f'{i}', end=' ')
    print(f'y={np.round(np.mean([np.mean(j) for j in i]),3)}')
    sample_2.append(i if i[1][0] >= 81 and i[0][0] < 81 else None)
```
