4. **Feature Enginering**

Feature engineering is transform the object or categorical variable data into float or integer data to be processes in exploratory data analysis and modelling or known as encode.

There are several technique in feature engineering, for instances:

1.   One hot encoding
2.   Mean encoding
3.   Frequency encoding

In this example, we will use frequency encoding since there ara many variations of value in categorical variable.
 



4.a Check number of variation value in "job" variable.

```python
print('Profession: ',data['Profession'].unique().size)
```

```python
data['Profession'].value_counts()
```

4.b Define frequency encoding function

```python
def features_encode(data,features):
    freq_encode=data.groupby(features).size()/len(data)
    return freq_encode
```

4.c Encode and mapping all categorical variable into frequency encoding

```python
#Encode
data.loc[:,'Gender_encode']=data['Gender'].map(features_encode(data,data['Gender']))
data.loc[:,'marital_encode']=data['Ever_Married'].map(features_encode(data,data['Ever_Married']))
data.loc[:,'education_encode']=data['Graduated'].map(features_encode(data,data['Graduated']))
data.loc[:,'Profession_encode']=data['Profession'].map(features_encode(data,data['Profession']))
data.loc[:,'Var_1_encode']=data['Var_1'].map(features_encode(data,data['Var_1']))
data.loc[:,'Spending_Score_encode']=data['Spending_Score'].map(features_encode(data,data['Spending_Score']))
```

6. **Split Data into Training and Testing**

```python
input=data_process.drop('Segmentation',axis=1)
output=data_process['Segmentation']
```

```python
x_train, x_test, y_train, y_test = train_test_split(input, output, test_size = 0.2, random_state = 123)
```

7. **Prediction Model**

```python
from sklearn.tree import DecisionTreeClassifier
dectree = DecisionTreeClassifier( random_state=123)
dectree.fit(x_train, y_train)
```

```python
y_predict=dectree.predict(x_test)
```

```python
x_test
```

```python
y_predict
```

8. **Calculate Performance of The Model**

```python
def score(y_test, y_pred):
    print("Confusion Matrx:")
    print(confusion_matrix(y_test, y_pred))
    print("Accuracy Score:", accuracy_score(y_test, y_pred))
    print("Recall Score:", recall_score(y_test,y_pred, average='macro'))
    print("Precision Score:", precision_score(y_test,y_pred,average='macro'))
    print("F1 - Score:", f1_score(y_test, y_pred,average='macro'))
```

```python
score(y_test,y_predict)
```

```python
print('\nClassification Report\n')
print(classification_report(y_test, y_predict, target_names=['A', 'B', 'C','D']))
```
