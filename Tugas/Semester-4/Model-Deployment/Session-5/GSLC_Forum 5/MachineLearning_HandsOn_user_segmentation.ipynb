4. **Split Data into Training and Testing**

```python
input_df=df.drop('Segmentation',axis=1)
output_df=df['Segmentation']
```

```python
x_train, x_test, y_train, y_test = train_test_split(input_df, output_df, test_size = 0.2, random_state = 42)
```

5. **Missing Value Handling**

```python
import statistics
```

```python
statistics.mode(x_train['Ever_Married'])
```

```python
statistics.mode(x_train['Graduated'])
```

```python
statistics.mode(x_train['Profession'])
```

```python
#Replace Missing Value for train data
x_train['Ever_Married'].fillna('Yes', inplace=True)
x_train['Graduated'].fillna('Yes', inplace=True)
x_train['Profession'].fillna('Artist', inplace=True)
```

```python
#Replace Missing Value for test data
x_test['Ever_Married'].fillna('Yes', inplace=True)
x_test['Graduated'].fillna('Yes', inplace=True)
x_test['Profession'].fillna('Artist', inplace=True)
```

```python
x_train.head()
```

```python
#Calculate Mean for Work Experience
np.mean(x_train['Work_Experience'])
```

```python
#Calculate Mean for Family Size
np.mean(x_train['Family_Size'])
```

```python
#Replace Null numeric value with their Mean for train data
x_train['Work_Experience'].fillna(2.7, inplace=True)
x_train['Family_Size'].fillna(2.9, inplace=True)
```

```python
#Replace Null numeric value with their Mean for train data
x_test['Work_Experience'].fillna(2.7, inplace=True)
x_test['Family_Size'].fillna(2.9, inplace=True)
```

```python
x_train.head()
```

```python
print(x_train.info())
```

6. **Feature Encoding**

```python
#encode non numeric data into numeric data
#encode Gender Graduateand ever married
train_encode={"Gender": {"Male":1,"Female" :0}, "Ever_Married":{"Yes":1, "No":0}, "Graduated":{"Yes":1, "No":0}}
test_encode={"Gender": {"Male":1,"Female" :0}, "Ever_Married":{"Yes":1, "No":0}, "Graduated":{"Yes":1, "No":0}}

#replace categorical value into numerical value
x_train=x_train.replace(train_encode)
x_test=x_test.replace(test_encode)
```

```python
x_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
#encoded profession and spending score for training data
x=x_train[['Profession','Spending_Score']]
train_encoded=OneHotEncoder()
train_data=pd.DataFrame(train_encoded.fit_transform(x).toarray(),columns=train_encoded.get_feature_names_out())
x_train=x_train.reset_index()
x_train_enc=pd.concat([x_train,train_data], axis=1)
```

```python
x_train_enc.head()
```

```python
#encoded profession and spending score for testing data
x=x_test[['Profession','Spending_Score']]
test_data=pd.DataFrame(train_encoded.transform(x).toarray(),columns=train_encoded.get_feature_names_out())
x_test=x_test.reset_index()
x_test_enc=pd.concat([x_test,test_data], axis=1)
```

```python
x_train_enc.columns
```

```python
x_train=x_train_enc[['Gender', 'Ever_Married', 'Age', 'Graduated', 
                     'Work_Experience', 'Family_Size','Profession_Artist', 
                     'Profession_Doctor', 'Profession_Engineer',
                     'Profession_Entertainment', 'Profession_Executive',
                     'Profession_Healthcare', 'Profession_Homemaker', 
                     'Profession_Lawyer','Profession_Marketing', 
                     'Spending_Score_Average', 'Spending_Score_High',
                     'Spending_Score_Low' ]]
```

```python
x_test=x_test_enc[['Gender', 'Ever_Married', 'Age', 'Graduated', 
                     'Work_Experience', 'Family_Size','Profession_Artist', 
                     'Profession_Doctor', 'Profession_Engineer',
                     'Profession_Entertainment', 'Profession_Executive',
                     'Profession_Healthcare', 'Profession_Homemaker', 
                     'Profession_Lawyer','Profession_Marketing', 
                     'Spending_Score_Average', 'Spending_Score_High',
                     'Spending_Score_Low' ]]
```

7. **Prediction Model**

```python
from sklearn.tree import DecisionTreeClassifier
dectree = DecisionTreeClassifier()
dectree.fit(x_train, y_train)
```

```python
y_predict=dectree.predict(x_test)
```

8. **Calculate Performance of The Model**

```python
print('\nClassification Report\n')
print(classification_report(y_test, y_predict, target_names=['A', 'B', 'C','D']))
```
