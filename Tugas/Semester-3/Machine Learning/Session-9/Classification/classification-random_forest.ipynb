```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./spotify_songs.csv')
df.head()
```

```python
# Check for dataset size and basic info
df.info()
```

```python
# Check for missing value
df.isnull().sum()
```

```python
# Drop Null Value
df.dropna(inplace=True)
df.isnull().sum()
```

```python
# Remove Useless Variables
df.drop(['track_id', 'track_name', 
         'track_album_id', 'track_album_name', 
         'track_album_release_date', 'track_artist', 
         'playlist_name', 'playlist_id'], axis=1, inplace=True)
df.info()
```

```python
# Check for outliers
def find_outliers(col):
    q1 = col.quantile(.25)
    q3 = col.quantile(.75)
    IQR = q3 - q1
    ll = q1 - (1.5*IQR)
    ul = q3 + (1.5*IQR)
    upper_outliers = col[col > ul].index.tolist()
    lower_outliers = col[col < ll].index.tolist()
    bad_indices = list(set(upper_outliers + lower_outliers))
    return(bad_indices)

import numpy as np
bad_indexes = []
for col in df.columns:
    if df[col].dtype in ["int64","float64"]:
        bad_indexes.append(find_outliers(df[col]))

bad_indexes = set(list(np.concatenate(bad_indexes).flat))
print(len(bad_indexes))
```

Ignore Outliers for now

```python
# Preprocessing
# One Hot Encoding
ohe_data = pd.get_dummies(df, columns=['playlist_genre'], dtype=int)
ohe_data.head()
```

```python
ohe_data.info()
```

## Classification

#### Split Data

```python
input_df = ohe_data.drop('playlist_subgenre', axis=1)
output_df = ohe_data['playlist_subgenre']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(input_df, output_df, test_size=0.25, random_state=1)
```

#### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
import warnings
warnings.filterwarnings('ignore')

RF_class = RandomForestClassifier(criterion= 'gini',max_depth=4, random_state=1)
RF_class.fit(x_train, y_train)
```

```python
predict_rf = RF_class.predict(x_test)
```

```python
from sklearn.metrics import classification_report

print('Classification Report')
print(classification_report(y_test, predict_rf))
```

#### CVGripSearch-Random Forest

```python
from sklearn.model_selection import GridSearchCV

parameters = {
    'criterion':['gini', 'entropy', 'log_loss'],
    'max_depth':[2,4,6,8,10], 
}
```

```python
RF_classCV = RandomForestClassifier(random_state=1)
RF_classCV = GridSearchCV(RF_classCV ,
                            param_grid = parameters,   # hyperparameters
                            scoring='accuracy',        # metric for scoring
                            cv=5)  
```

```python
RF_classCV.fit(x_train,y_train)
print("Tuned Hyperparameters :", RF_classCV.best_params_)
print("Accuracy :",RF_classCV.best_score_)
```

```python
RF_class_best = RandomForestClassifier(criterion= 'gini',max_depth=10, random_state=1)
```

```python
RF_class_best.fit(x_train,y_train)
```

```python
y_predict_best=RF_class_best.predict(x_test)
```

```python
print('\nClassification Report\n')
print(classification_report(y_test, y_predict_best))
```
