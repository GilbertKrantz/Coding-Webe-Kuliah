```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

## Import Data

```python
df = pd.read_csv('./student_data.csv')
df.head()
```

## Check data type

```python
df.info()
```

## Check Missing Value

```python
df.isnull().sum()
```

## Encode and drop useless variables

```python
new_df = df.drop(['Student Name', 'Date of Birth', 'Student ID'], axis=1)
new_df
```

```python
from sklearn.preprocessing import LabelEncoder
# instantiate labelencoder object
le = LabelEncoder()

# apply le on categorical feature columns
new_df['Specialization'] = le.fit_transform(new_df['Specialization'])
new_df['Field of Study'] = le.fit_transform(new_df['Field of Study'])
```

## Split Data

```python
X = new_df.drop('Discount on Fees', axis=1)
y = new_df['Discount on Fees']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

## Regression

#### Linear

```python
ln_regression = LinearRegression()
ln_regression.fit(X_train, y_train)
```

```python
predict_ln = ln_regression.predict(X_test)
print("r2 :", r2_score(y_test, predict_ln))
print("MSE :", mean_squared_error(y_test, predict_ln))
```

#### KNN

```python
knn_regression = KNeighborsRegressor()
knn_regression.fit(X_train, y_train)
```

```python
predict_knn = knn_regression.predict(X_test)
print("r2 :", r2_score(y_test, predict_knn))
print("MSE :", mean_squared_error(y_test, predict_knn))
```

#### Decission Tree

```python
dt_regression = DecisionTreeRegressor()
dt_regression.fit(X_train, y_train)
```

```python
predict_dt = dt_regression.predict(X_test)
print("r2 :", r2_score(y_test, predict_dt))
print("MSE :", mean_squared_error(y_test, predict_dt))
```
