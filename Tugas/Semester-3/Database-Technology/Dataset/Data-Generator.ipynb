```python
import pandas as pd
from faker import Faker
```

```python
faker = Faker()
```

```python
def generator() :
    # Seller Data
    seller_id = f'SELLER_{faker.random_number(digits=5)}'
    seller_name = faker.name()
    seller_domicile = faker.address()
    seller_phone = faker.phone_number()
    seller_dob = faker.date_of_birth()

    # House Data
    house_id = f'HOUSE_{faker.random_number(digits=5)}'
    house_address = faker.address()
    house_land_area = faker.random_number(digits=4)
    house_building_area = faker.random_number(digits=4) + house_land_area*2
    house_pool = faker.boolean()
    house_bedroom = faker.random_number(digits=1)
    house_bathroom = faker.random_number(digits=1)
    house_floor = faker.random_number(digits=1)
    house_condition = faker.random_element(elements=('New', 'Second', 'Renovated'))
    house_price = faker.random_number(digits=9) + house_land_area*100 + house_building_area*100
    house_province = faker.state()
    house_city = faker.city()

    # Buyer Data
    buyer_id = f'BUYER_{faker.random_number(digits=5)}'
    buyer_name = faker.name()
    buyer_domicile = faker.address()
    buyer_phone = faker.phone_number()
    buyer_dob = faker.date_of_birth()

    # Transaction Data
    transaction_id = f'TRANSACTION_{faker.random_number(digits=5)}'
    transaction_date = faker.date()
    transaction_price =  house_price

    # Linking
    seller_house_id = house_id
    buyer_house_id = house_id
    transaction_seller_id = seller_id
    transaction_buyer_id = buyer_id
    transaction_house_id = house_id

    # list
    seller = [seller_id, seller_name, seller_domicile, seller_phone, seller_dob]
    house = [house_id, house_address, house_land_area, house_building_area, house_pool, house_bedroom, house_bathroom, house_floor, house_condition, house_price, house_province, house_city]
    buyer = [buyer_id, buyer_name, buyer_domicile, buyer_phone, buyer_dob]
    transaction = [transaction_id, transaction_date, transaction_price]

    return seller, house, buyer, transaction

```

```python
generator()
```

```python
# Make Dataframe
seller = pd.pandas.DataFrame(columns=['seller_id', 'seller_name', 'seller_domicile', 'seller_phone', 'seller_dob'])
house = pd.pandas.DataFrame(columns=['house_id', 'house_address', 'house_land_area', 'house_building_area', 'house_pool', 'house_bedroom', 'house_bathroom', 'house_floor', 'house_condition', 'house_price', 'house_province', 'house_city'])
buyer = pd.pandas.DataFrame(columns=['buyer_id', 'buyer_name', 'buyer_domicile', 'buyer_phone', 'buyer_dob'])
transaction = pd.pandas.DataFrame(columns=['transaction_id', 'transaction_date', 'transaction_price'])
```

```python
# Generate Data
for i in range(50000) :
    seller_data, house_data, buyer_data, transaction_data = generator()
    seller.loc[i] = seller_data
    house.loc[i] = house_data
    buyer.loc[i] = buyer_data
    transaction.loc[i] = transaction_data
```
