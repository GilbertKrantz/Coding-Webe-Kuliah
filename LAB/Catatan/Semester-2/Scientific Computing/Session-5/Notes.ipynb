```python
import numpy as np
```

```python
def f(x):
  return x**2 + 3

# hitung luas kurva dari a->b
a = 2
b = 5
n = 5 # n = jmlh kotak + 1 (krn yg dihitung titik dibawah)

lebar_kotak = (b - a) / (n - 1)
x = np.linspace(a, b, n)
# dapatkan n titik dari a-b dgn interval n
# a, b, n
# a = mulai dari angka brp
# a = berakhir di angka brp
# n = interval
print(f'x = {x}')

y = f(x)
print(f'y = {y}')

leftRiemann = lebar_kotak * sum(y[:n-1]) # y di sum mulai dr awal sampai angka terakhir kurang 1
rightRiemann = lebar_kotak * sum(y[1:]) # y di sum mulai dr idex ke 1 sampai akhir

print(f'Left Riemann = {leftRiemann}')
print(f'Right Riemann = {rightRiemann}')


x_mid = (x[:n-1] + x[1:]) / 2
# print(x[:n-1])
# print(x[1:])
# print(x_mid)
y_mid = f(x_mid)
mid = lebar_kotak * sum(y_mid)
print(f'Mid Riemann = {mid}')
```
