# Regression

### Import Library

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
```

### Initialize Array

```python
x = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([2, 5, 1, 7, 9, 3, 6, 10])
```

## Least Square Regression
Usually for Linear Graph

```python
A = np.vstack([x, np.ones(len(x))]).T
# print(A)

# Make Y into 2 d Array
Y = y[:, np.newaxis]
# print(Y)

# Pseudo Inverse
result = np.dot(np.linalg.pinv(A), Y)

# Direct inverse
# result = np.dot(np.dot(np.linalg.inv(np.dot(A.T, Y)), A.T), Y)
# print(result)

m = result[0]
b = result[1]

# y = mx + b

plt.plot(x, m * x + b, 'b') # Line
plt.plot(x, y, 'ro')

plt.show()
```

## Polynomial Regression
Usually for Non-Linear Graph

```python
plt.figure(figsize=(16, 12))    
for i in range(1, 7) :
    plt.subplot(2, 3, i)
    plt.plot(x, y, 'bo')
    
    # Search for the Polynomial Coeficient
    y_est = np.polyfit(x, y, i)
    
    # Plotting
    plt.plot(x, np.polyval(y_est, x), 'r') 
    plt.title(f'Polynomial Regression of {i}')

plt.show()
```

## Cubics Spline Interpolation
Estimate x and y from the data that we already have

```python
x = np.array([0, 1, 2])
y = np.array([1, 3, 2])

# Make CubicSpline Function
f = CubicSpline(x, y)

# Generate 100 data from the range of 0 - 2 with the same interval
x_new = np.linspace(0, 2, 100)

y_new = f(x_new)

plt.plot(x_new, y_new, 'r')
plt.plot(x, y, 'bo')
plt.title("Cubic Spline Interpolation")
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
