# Root Finding
f(x) = x^2 - 4

Find the value of x when y - 0
0 = x^2 - 4
4 = x^2
x = 2

```python
import numpy as np
```

## Bisection Method
Concept -> Find unique value.
Requirements -> f(a) and f(b) can't be positive(+) or negative(-) at the same time.
Formula Mid = (a + b) / 2

```python
# Define Equation
def f(x) :
    return x**2 - 4

# Define Bisection Function
def bisection(a, b, tolerance = 0.01) :
    # Validate Point
    # np.sign -> check value is positive or negative
    if np.sign(f(a)) == np.sign(f(b)):
        print("Invalid Coordinate")
        return
    # Valid Coordinate

    # Find Mid Value
    mid = (a + b) / 2

    # 3 Condition
    # abs(f(mid)) < tolerance
    if np.abs(f(mid)) < tolerance :
        # Coordinate Found
        print(f"Coordinate Found At : {mid}")
        return
    # if f(mid) =- f(b)
    elif np.sign(f(mid)) == np.sign(f(b)) :
        bisection(a, mid, tolerance)
    elif np.sign(f(mid)) == np.sign(f(a)) :
        bisection(mid, b, tolerance)

bisection(-4,-3,0.01) # Invalid Coordinate
bisection(-3,-1,0.01) # Valid Coordinate
```

## Newton-Raphson Method
Equation
f(x) = 3x^3 - 2x - 1

Derivative
g(x) = 9x^2 - 2

Formula
Xnew = x - (f(x) / g(x))
If abs(f(Xnew)) < tolerance -> **Found**


```python
# Base Equation
def f(x) :
    return 3*x**3 - 2*x - 1

def g(x) :
    return 9*x**2 - 2

def newton_raphson(x, tolerance = 0.01) :
    # Enter Formula from Xnew = x - (f(x) / g(x))
    x_new = x - (f(x) / g(x))
    if np.abs(f(x_new)) < tolerance :
        print(f"Root Found at {x_new}")
        return
    newton_raphson(x_new, tolerance)

newton_raphson(100)
```
