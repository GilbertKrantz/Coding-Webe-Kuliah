```python
import numpy as np
```

```python
# Variable Intialization

Xs = np.array([
    # Soal 1
    [
        [10, 4, 5],
        [1, 6, 1],
        [3, 1, 6]
    ],
    # Soal 2
    [
        [9, 2, 3, 2],
        [4, 11, 2, 3],
        [1, -4, 8, -1],
        [5, 3, -6, 16]
    ]
], dtype=object)

Ys = np.array([
    # Soal 1
    [8, 5, 10],
    # Soal 2
    [12, 9, 6, 21]
], dtype=object)
```

```python
# Gauss Seidel Function

def gauss_seidel(x, y, epsilon = 0.01, t = 20) :
    # Change List to array
    x = np.array(x)
    y = np.array(y)

    # Get Absolute Diagonal

    diags = np.diag(np.abs(x))

    # Fill Diagonal with 0
    np.fill_diagonal(x, 0)

    # Find the Value of not diagonal
    off_diags = np.sum(np.abs(x), axis = 1)

    # Check the diagonal dominance
    if not np.all(diags > off_diags) :
        return False
    
    # Change all X to negative
    x = -x

    # Intialize zeros matrix
    mat_old = np.zeros(x[0].shape)

    # Iteration
    # First Iteration
    for i in range(t) :
        # Copy mat_old to mat_new
        mat_new = np.array(mat_old)

        # Second Iteration
        for j, row in enumerate(x) :
            mat_new[j] = (y[j] + np.dot(mat_new, row)) / diags[j]

        # Displaying Variable data for eaxh iteration
        print(f"iter {i} : {mat_new}")

        # Calculation Euclidean Distance
        dx = np.sqrt(np.dot(mat_new - mat_old, mat_new - mat_old))
        if dx < epsilon :
            return True
        
        # Update mat_old
        mat_old = mat_new
    
    return False
```

```python
# Printing array

for i, (x, y) in enumerate(zip(Xs,Ys)) :
    print(f"Problem {i + 1}")
    print(f"x : {x}")
    print(f"y : {y}")

    if gauss_seidel(x, y) :
        print("Convergen")
    else :
        print("Not Convergen")
```

```python
# Gauss_Seidel

def Gauss_Seidel(x, y, epsilon = 0.01, t = 20) :
    # Initialize Array
    x = np.array(x)
    y = np.array(y)

    # Get diags
    diags = np.diag(np.abs(x))

    # Set all diagonal to 0
    np.fill_diagonal(x, 0)

    # Calculate off diags
    off_diags = np.sum(np.abs(x), axis=1)

    # Check Diagnoal Dominance
    if not np.all(diags > off_diags) :
        return False
    
    # Set x to negative
    x = -x

    # Make new zeros matrix
    old_mat = np.zeros(x[0].shape)

    # Iteration
    for i in range(t) :
        # Copy old Matrix to new Matrix
        new_mat = np.array(old_mat)

        # Second Iteration
        for j, row in enumerate(x) :
            new_mat[j] = (y[j] + np.dot(new_mat, row)) / diags[j]

        # Print Iteration
        print(f"Iter {i} : {new_mat}")
        
        # Calculate Euclidean Distance
        dx = np.sqrt(np.dot(new_mat - old_mat, new_mat - old_mat))
        if dx < epsilon :
            return True
        
        # Copy new_mat to old_mat
        old_mat = new_mat
    
    return False
```

```python
# Print
for i, (x, y) in enumerate(zip(Xs, Ys)) :
    print(f"Problem {i + 1}")
    print(f"x = {x}")
    print(f"y = {y}")

    if Gauss_Seidel(x, y) :
        print("Convergen")
    else :
        print("Not Convergen")
```
