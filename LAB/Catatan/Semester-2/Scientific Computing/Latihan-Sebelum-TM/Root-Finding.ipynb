```python
# Bisection
# Define Equation

def f(x) :
    return 2*x**3 + x**2 - x + 2

def bisection(a, b, tolerance = 0.01) :
    # Check for sign
    if np.sign(f(a)) == np.sign(f(b)) :
        print(f"Invalid Coordinate")
        return
    
    print(f"[{a}, {b}]\n")
    print(f"f({a}) = 2({a})^3 + ({a})^2 - ({a}) + 2 = {f(a)}")
    print(f"f({b}) = 2({b})^3 + ({b})^2 - ({b}) + 2 = {f(b)}")
    
    # Get Mid Value
    mid = (a + b) / 2
    print(f"mid = (({a}) + ({b})) / 2 = {mid}")

    print(f"f({mid}) = 2({mid})^3 + ({mid})^2 - ({mid}) + 2 = {f(mid)}\n")

    # If Mid value is smaller than tolerancew
    if np.abs(f(mid)) < tolerance :
        print(f"Coordinate Found : {mid}")
        return
    elif np.sign(f(mid)) == np.sign(f(b)) :
        bisection(a, mid, tolerance)
    elif np.sign(f(mid)) == np.sign(f(a)) :
        bisection(mid, b, tolerance)

# bisection(-4,-3,0.01) # Invalid Coordinate
# bisection(-3,-1,0.01) # Valid Coordinate
bisection(-2,0,0.0001) # Valid Coordinate
```

```python
# Bisection
def f(x) :
    return 2*x**3 + x**2 - x + 2

def bisection(x, y, tolerance = 0.01) :
    
    if np.sign(f(x)) == np.sign(f(y)) :
        print("Invalid Coordinate")
        return
    
    mid = (x + y) / 2

    if np.abs(f(mid)) < tolerance :
        print(f"Coordinate Found : {mid}")
        return
    elif np.sign(f(mid)) == np.sign(f(y)) :
        bisection(x, mid, tolerance)
    elif np.sign(f(mid)) == np.sign(f(x)) :
        bisection(mid, y, tolerance)

bisection(-2,0,0.0001) # Valid Coordinate
```

```python
# Newton Raphson

# Declare Equation
def f(x) :
    return 2*x**3 + x**2 - x + 2

def g(x) :
    return 6*x**2 + 2*x - 1

print(f"f(x) = 2x^3 + x^2 - x + 2")
print(f"g(x) = 6x^2 + 2x - 1\n")

def newtonRaphson(x, tolerance = 0.01) :
    x_new = x - f(x) / g(x)
    print(f"x_new = {x} - (({f(x)}) / ({g(x)})) = {x_new}")
    print(f"f(x_new) = {f(x_new)}\n")

    if np.abs(f(x_new)) < tolerance :
        print(f"Root Found at : {x_new}")
        return
    newtonRaphson(x_new, tolerance)

newtonRaphson(-2, 0.0001)
```

```python
# Newton Raphson

# Declare Equation
def f(x) :
    return 2*x**3 + x**2 - x + 2

def g(x) :
    return 6*x**2 + 2*x - 1

def newton_raphson(x, tolerance = 0.01) :
    x_new = x - (f(x) / g(x))

    if np.abs(f(x_new)) < tolerance :
        print(f"Root Found : {x_new}")
        return
    else :
        newton_raphson(x_new, tolerance)

newton_raphson(-2, 0.0001)
```
