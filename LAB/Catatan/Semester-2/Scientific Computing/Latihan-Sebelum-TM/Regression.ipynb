```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline
```

```python
x = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([2, 5, 1, 7, 9, 3, 6, 10])
```

```python
# Least Square Regression

# Make A 
A = np.vstack([x, np.ones(len(x))]).T

# Make Y into 2d Array
Y = y[:, np.newaxis]

# Calculate Pseudo Inverse
result = np.dot(np.linalg.pinv(A), Y)

m = result[0]
b = result[1]

plt.plot(x, m * x + b, 'b')
plt.plot(x, y, 'ro')

plt.show
```

```python
# Polynomial Regression

plt.figure(figsize=(16,14))

for i in range(1, 7) :
    plt.subplot(2, 3, i)
    plt.plot(x, y, 'ro')

    # Search for Polunomial Coefficient
    y_est = np.polyfit(x, y, i)

    plt.plot(x, np.polyval(y_est, x), 'b')

plt.show
```

```python
# Cubic Spiline Interpolation

f = CubicSpline(x, y)

x_new = np.linspace(1, 8, 100)

y_new = f(x_new)

plt.plot(x_new, y_new, 'r')
plt.plot(x, y, 'bo')
plt.show
```
