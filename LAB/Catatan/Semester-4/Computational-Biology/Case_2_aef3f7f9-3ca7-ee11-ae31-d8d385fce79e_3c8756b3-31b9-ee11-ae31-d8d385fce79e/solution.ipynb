```python
from Bio.Seq import Seq
import Bio.SeqUtils as Utils
import matplotlib.pyplot as plt
```

```python
seqA = Seq("AGCTTGCAGCGTCCGTTAGCTCGAGTCCAGGACGTTAGTCCTGCAGTC")
seqB = Seq("CAGTAAGTTGCCGTTAGCGCGTAGTGCCAGTAAGCGGCTCGTTAGTGG")
```

# 1. Molecular Weight of both sequences

```python
print(Utils.molecular_weight(seqA))
print(Utils.molecular_weight(seqB))
```

# 2. Refer to the Wallace rule to calculate the melting point of the sequence A and B and plot

```python
from Bio.SeqUtils import MeltingTemp as mt
```

```python
wallaceA = mt.Tm_Wallace(seqA)
wallaceB = mt.Tm_Wallace(seqB)
```

```python
plt.pie([wallaceA, wallaceB], labels=[wallaceA, wallaceB])
plt.legend(['Sequence A', 'Sequence B'])
```

# 3 and 4 Calculate the GC and AT content of the sequence A and B and plot

```python
GC_A = Utils.GC(seqA)
AT_A = 100 - GC_A
GC_B = Utils.GC(seqB)
AT_B = 100 - GC_B

print(f'Sequence A : GC({GC_A}), AT({AT_A})')
print(f'Sequence B : GC({GC_B}), AT({AT_B})')
```

# 4. Plot the GC and AT content of the sequence A and B

```python
plt.pie([GC_A, AT_A], labels=[GC_A, AT_A])
plt.legend(['GC', 'AT'])
plt.show
```

```python
plt.pie([GC_B, AT_B], labels=[GC_B, AT_B])
plt.legend(['GC', 'AT'])
plt.show
```
