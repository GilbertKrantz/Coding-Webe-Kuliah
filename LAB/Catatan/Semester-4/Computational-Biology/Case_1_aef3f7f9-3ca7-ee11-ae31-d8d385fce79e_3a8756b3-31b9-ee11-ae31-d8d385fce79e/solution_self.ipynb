```python
from Bio.Seq import Seq
from Bio.SeqUtils import nt_search
from collections import Counter
import matplotlib.pyplot as plt
```

```python
seqA = Seq('AGCTTGCAGCGTCCGTTAGCTCGAGTCCAGGACGTTAGTCCTGCAGTC')
seqB = Seq('CAGTAAGTTGCCGTTAGCGCGTAGTGCCAGTAAGCGGCTCGTTAGTGG')
```

```python
print(len(seqA))
print(len(seqB))
```

```python
print(seqA.count('CGC'))
print(seqB.count('CGC'))
```

```python
print(nt_search(str(seqA), 'CAGTC'))
print(nt_search(str(seqB), 'CAGTC'))
```

```python
seqC = seqA[:15] + seqB[-10:]
seqC
```

```python
seqC[::-1]
```

```python
def plot_nucleotide(Sequence) :
    dna_coll = Counter(Sequence)
    plt.bar(dna_coll.keys(), dna_coll.values())
    
plot_nucleotide(seqA) 
plot_nucleotide(seqB) 
plot_nucleotide(seqC) 

plt.show()
```
