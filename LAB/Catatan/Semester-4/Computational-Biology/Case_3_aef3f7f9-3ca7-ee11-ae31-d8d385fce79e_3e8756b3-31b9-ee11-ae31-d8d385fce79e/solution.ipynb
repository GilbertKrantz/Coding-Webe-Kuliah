```python
from Bio.Seq import Seq
import Bio.SeqUtils as Utils
```

```python
seqA = Seq('ATGCGTTTATGACGTATCTGAAGTCTACTAGGGTGATCTGACTAA')
seqB = Seq('GTGTTAGTGCGCTAGTGCTGACTAGGTACAGTAGCTAGTTACTAG')
```

# 1. Transcribe the DNA into RNA 

```python
# Complement
seqA_Complement = seqA.complement()
seqB_Complement = seqB.complement()

# Transcribe
seqA_mRNA = seqA.transcribe()
seqB_mRNA = seqB.transcribe()
```

```python
print(seqA_mRNA)
print(seqB_mRNA)
```

# 2. Combine the 24 starting nucleotides of mRNA sequence A and 21 ending nucleotides of mRNA sequence B to form a new mRNA sequence C

```python
seqC_mRNA = seqA_mRNA[:24] + seqB_mRNA[-21:]
seqC_mRNA
```

# 3. Translate the mRNA sequence A, B, C into a protein sequence

```python
seqA_amino = seqA_mRNA.translate()
seqB_amino = seqB_mRNA.translate()
seqC_amino = seqC_mRNA.translate()
```

```python
print(seqA_amino)
print(seqB_amino)
print(seqC_amino)
```

# 4. Transcribe back sequence C into DNA

```python
seqC = seqA_mRNA.back_transcribe()
seqC
```

# 6. Calculate the Complement of the DNA sequence

```python
seqA_Complement = seqA.complement()
seqB_Complement = seqB.complement()
seqC_Complement = seqC.complement()
```

```python
print(seqA_Complement)
print(seqB_Complement)
print(seqC_Complement)
```
