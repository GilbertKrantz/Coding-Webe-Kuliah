```python
from Bio.Seq import Seq
import Bio.SeqUtils as Utils
```

```python
seqA = Seq('AAGCTTAGCTAGAGCAGTCAGCTAGCTAGCT')
seqB = Seq('TCTGAAGCTAGCTAGTGCATAGCTAGCTAGC')
```

# 1. Determine score of best local allignment

```python
from Bio import pairwise2 as pw2
```

```python
local_alignments = pw2.align.localxx(seqA, seqB)
```

```python
for aligments in local_alignments :
    print(aligments)
    print(pw2.format_alignment(*aligments))
    
```

```python
global_alignment = pw2.align.globalms(seqA, seqB, 1, -1, -1, -1)
for x in global_alignment:
    print(pw2.format_alignment(*x))
```

# 2. Hamming Distance Similarity

```python
def hamming(seq1, seq2) :
    h = []
    for x, y in zip(seq1, seq2):
        if x != y :
            h.append((x,y))
    return len(h)

hamming(seqA, seqB)
```

# 3. Levinshtein Distance Similarity

```python
from Levenshtein import distance

print(distance(seqA, seqB))
```

# 4. Dot Plot

```python
def delta(x, y) :
    if x != y : return 1 
    else : return 0
    
def M(seq1, seq2, seq1_start, seq2_start, target_length):
    m = []
    
    for x, y in zip(seq1[seq1_start:target_length], seq2[seq2_start:target_length]) :
        m.append(delta(x, y))
    
    return sum(m)

def make_matrix(seq1, seq2, target_length):
    m1 = []
    for i in range(len(seq1) - target_length + 1):
        m2 = []
        for j in range(len(seq2) - target_length + 1):
            m2.append(M(seq1, seq2, i, j, target_length))
        m1.append(m2)
    return m2

def plot_mat(M, tolerance, seq1, seq2, nonBlank=chr(0x25A0), blank=" "):
    print(' |' + seq2)
    print('-' * (4 + len(seq2)))
    
    for label, row in zip(seq1, M):
        line = ''.join(nonBlank if score < tolerance else blank for score in row)
        print(label + '|' + line)
        
def dot_plot(seq1, seq2, k=1, t=1):
    matrix = make_matrix(seq1, seq2, k)
    plot_mat(matrix, t, seq1, seq2)
```

```python
dot_plot(seqA, seqB)
```
