```python
list = [1, 2, 3, 3]

unique_list = []

for ele in list :
    count = 0
    print(ele, end=" ")
    for elem in list : #Looping dalam looping
        if ele == elem :
            count = count + 1
        print(elem, end=" ")
    if count == 1 :
        unique_list.append(ele)
    print("\n")

print(unique_list)
```
