Cara 1 (List Comprehension)

```python
list_of_dict = [dict(zip(["Name", "GPA", "Study Program", "SAT Point"], vals)) for vals in zip(Name, GPA, Study_Program, SAT_Point)] 
list_of_dict
```

Cara 2 (Looping)

```python
list_of_dict = []

for i in range(len(Name)) :
    dict = {"Name" : Name[i], "GPA": GPA[i], "Study Program": Study_Program[i], "SAT Point": SAT_Point[i]}
    list_of_dict.append(dict)

list_of_dict
```

Nomor 3

```python
#  Cari mean dari GPA semua anak data science

sum = 0 # Total GPA
sumLength = 0 # Total Panjang GPA

# Format : list_of_dict[index][key] 

for i in range(len(list_of_dict)) :
    if list_of_dict[i]["Study Program"] == "Data Science" :
        sum = sum + list_of_dict[i]["GPA"]
        sumLength = sumLength + 1

print(sum, sumLength)

mean = sum / sumLength
print(mean)
```
