### Even-Odd Sum

```python
list = [2, 57, 18, 60, 17, 1, 6, 59, 19, 64, 87, 17, 69, 83, 92, 55, 5, 12, 22, 73, 25, 83]
sum = 0

for i in list :
    if (i % 2 == 0) :
        sum = sum + i

print(sum)
```

### Finding Unique Elements

```python
list = []

list_len = int(input("List Length: "))
for i in range(list_len) :
    element = int(input())
    list.append(element)

print(list)
```

```python
unique_list = []

for i in list :
    occur = 0
    for j in list :
        if (j == i) :
            occur = occur + 1
    
    if (occur <= 1) :
        unique_list.append(i)

print(unique_list)
```

### Factorial Calculator

```python
val = int(input("Input a Number: "))

factorial_mult = 1
factorial = 1

for i in range(val) :
    factorial = factorial * factorial_mult
    factorial_mult = factorial_mult + 1

print(factorial)
```

### Pattern Printer

```python
rows = int(input("Rows : "))

for i in range(rows) :
    if (i == 0) :
        print("*")
    elif (i == rows - 1) :
        for j in range(i+1) :
            print("*", end='')
    else :
        print("*", end='')
        for j in range(i - 1) :
            print(" ", end='')
        print("*")
```

### Palindrome

```python
str = input("String : ")

palindrome = ''
for i in range(len(str)) :
    j = -1 + (-1 * i)
    palindrome = palindrome + str[j]

if palindrome == str :
    print(f"{str} is a Palindrome")
else :
    print(f"{str} isn't a Palindrome")
```

## Function

### Circle Area Calculator

```python
pi = 3.14
def circle_area(radius) :
    return (radius**2) * pi

circle_area(2)
```

### String Reverser

```python
def str_reverse(string) :
    reversed = ''
    for i in range(len(string)) :
        reversed = reversed + string[-1 + (i * -1)]
    return reversed

str_reverse("My Name is Wilbert")
```

### Exponent Calculator

#### With Loop

```python
def exp_calc_loop(a, b) :
    total = a
    for i in range(b-1) :
        total = total * a
    return total

exp_calc_loop(3,3)
```

#### Without Loop

```python
def exp_calc_no_loop(a, b) :
    return a ** b

exp_calc_no_loop(2, 3)
```

## Indexing & Slicing

### Middle Elements

```python
list = ["Dragon", "Fire", "Water", "Snake"]

def mid_extract(list) :
    middle = (len(list)-1)/2
    if (len(list) % 2 != 0) :
        print(list[round(middle)])
    else :
        print(f"{list[int(middle - 0.5)]} or {list[int(middle + 0.5)]}")

mid_extract(list)
```

### Vowel Extractor

```python
def vowel_extract(string) :
    list = []
    for char in string :
        if (char == 'a') or (char == 'i') or (char == 'u') or (char == 'e') or (char == 'o') :
            list.append(char)
    return(list)

vowel_extract("Wilbert")
```

### List Reverser

```python
def list_reverse(list) :
    reversed = []
    for i in range(len(list)) :
        reversed.append(list[-1 + (i * -1)])
    return reversed

list = [2, 3, 4]
list_reverse(list)
```

## Basic Operation

### Calculator Function

```python
def calculator(a, b, operator) :
    if operator == '+' :
        return a + b
    elif operator == '-' :
        return a - b
    elif operator == '*' or operator == 'x':
        return a * b
    elif operator == '/' :
        if b == 0 :
            return "Can't be divided"
        else :
            return a / b
    else :
        return "Wrong Operator Input"
    
calculator(1, 0, '/')
```

### Simple Interest Calculator

```python
def interest_calc(principal, rate, time) :
    for i in range(time) :
        principal = principal + (principal * rate)
    return principal

interest_calc(1000, 0.2, 3)
```

```python
def fibonacci(N) :
    list = [0, 1]
    for i in range(N - 2) :
        sum = list[i] + list[i + 1]
        list.append(sum)
    return list

fibonacci(10)
```
