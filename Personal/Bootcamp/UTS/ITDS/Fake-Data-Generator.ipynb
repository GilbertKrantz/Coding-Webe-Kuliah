```python
from faker import Faker
import pandas as pd
```

```python
fake = Faker()
```

```python
def generator() :
    i = fake.random_int(min=1, max=9999)
    # Create Unique ID with Format : JKT-0001
    unique_id = f'JKT-{i:04}'
    # Create Fake Data Consisting of Age, Gender, Math Score, Reading Score, Writing Score, and Lunch Type (Standard or Free/Reduced).
    age = fake.random_int(min=15, max=18)
    Gender = fake.random_element(elements=('Male', 'Female'))
    Math_Score = fake.random_int(min=0, max=100)
    Reading_Score = fake.random_int(min=0, max=100)
    Writing_Score = fake.random_int(min=0, max=100)
    Lunch = fake.random_element(elements=('Standard', 'Free/Reduced'))

    # Create a List to Store the Data
    data = [unique_id, age, Gender, Math_Score, Reading_Score, Writing_Score, Lunch]
    return data
```

```python
generator()
```

```python
# Create a dataframe to store the data
df = pd.DataFrame(columns=["unique_id", "age", "Gender", "Math_Score", "Reading_Score", "Writing_Score", "Lunch"])
df
```

```python
for i in range(10000):
    df.loc[i] = generator()
```

```python
# Save the dataframe to csv file
df.to_csv('fake-data.csv', index=False)
```
