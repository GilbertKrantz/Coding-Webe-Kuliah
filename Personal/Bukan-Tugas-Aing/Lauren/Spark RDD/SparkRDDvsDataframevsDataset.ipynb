# Spark RDD vs Dataframe vs Dataset

## Spark RDD

### Creating RDD

#### From Existing Collection

```python
from pyspark import SparkContext

# Initialize Spark Context
sc = SparkContext("local", "RDD Exanple")

# Create a list
data = [1, 2, 3, 4, 5]

# Parallelize the collection to create an RDD
rdd = sc.parallelize(data)

print(rdd.collect()) # Output: [1, 2, 3, 4, 5]
```

#### From External Data Source

```python
from pyspark import SparkContext

# Initialize Spark Context
sc = SparkContext("local", "RDD Exanple")

# Read data from a csv file to create an RDD
rdd = sc.textFIle("movie_data_part1.csv")

# Print the first 4 lines
print(rdd.take(5))
```

### Basic RDD Operation

#### Transformation

```python
from pyspark import SparkContext

# Initialize Spark Context
sc = SparkContext("local", "RDD Exanple")

# Create a list
data = [1, 2, 3, 4, 5]
rdd = sc.parallelize(data)

# Apply a map transformation (e.g., multiply each element by 2)
mapped_rdd = rdd.map(lambda x: x * 2)

# Apply a filter transformation (e.g., keep even numbers)
filtered_rdd = mapped_rdd.filter(lambda x: x % 4 == 0)
```

#### Action

```python
result = filtered_rdd.collect()
print(result)

# Count the number of elements in the original RDD
count = rdd.count()
print(count)
```

## Spark Dataframe

### Creating a Spark Dataframe

```python
from pyspark.sql import SparkSession
from pyspark.sql import Row

# Initialize SparkSession
spark=SparkSession.builder.appName("Data_Wrangling").getOrCreate()

# Create an RDD of Rows
data = [Row(name="Alice", age=25), Row(name="Bob", age=30), Row(name="Charlie", age=35)]
rdd = spark.sparkContext.parallelize(data)

# Convert RDD to Dataframe
df = spark.createDataFrame(rdd)

# Stop the SparkSession
spark.stop()
```

### Basic Operation

```python
df.show() # Display top 20 rows
df.printSchema() # Display schema of the DataFrame
```

```python
# Filter rows where age > 30
df.filter(df.age > 30).show()
```

```python
# Select the "name" and "age" columns
df.select("name", "age").show()
```

```python
# Read data
file_location = "movie_data_part1.csv"
file_type = "csv"
infer_schema = "false"
first_row_is_header = "true"
delimiter = "|"

df = spark.read.format(file_type)\
.option("inferSchema", infer_schema)\
.option("header", first_row_is_header)\
.option("sep", delimiter)\
.load(file_location)
```
