```python
clothes = {
    "Shirt" : 150000,
    "T-shirt" : 100000,
    "Trousers" : 200000,
    "Shorts" : 120000,
    "Jackets" : 250000
}
```

```python
Total = 0
index = list()


print(f"Enter the of clothes you want to buy :", end=" ")
amount = int(input())

print(f"Available Clothes : ")
for x, i in enumerate(clothes) :
    print(f"{x} - {i}")

for i in range(amount) :
    print(f"Enter the index of the type of clothes : ", end=" ")
    print("")
    index.append(int(input()))
    Total = Total + list(clothes.values())[index[i]]

print(f"\n------- Invoice -------")
print(f"Clothes Purchased : ")
for i in range(amount) :
    print(f"{list(clothes)[index[i]]}")

discount = 0.1 if (Total >= 1000000) else 0.05 if (Total >= 500000) else 0

print(f"Total price before discount : {Total}")
print(f"Discount : {discount * 100}%")
print(f"Total price after discount : {Total - Total * discount}")
```
