```python
def is_prime(input_num) :
    if input_num <= 1 :
        return False
    elif input_num <= 3 : 
        return True
    elif input_num % 2 == 0 or input_num % 3 == 0 :
        return False
    
    return all(input_num % i for i in range(5, int(np.sqrt(input_num)) + 1, 2))

def prime_spitter(input_limit) :
    for num in range(input_limit) :
        if is_prime(num) :
            print(num, end=", ")

prime_spitter(100)
```

Explanation

Function is_prime untuk mecari tahu angka tsb prima atau tidak. Hal ini dilakukan dengan me return true angka 2 dan 3 terlebih dahulu serta me return false angka yang habis dibagi 2 atau 3. setelah itu kita mencari apakah angka tersebut prima dengan menghitung modulus bilangan input dengan i, dimana i adalah suatu angka dari iterasi range 5 sampai integer akar bilangan input + 1 dengan step 2.

Functoin prime_spitter untuk memuntahkan atau meprint angka yang dikenal sebagai prima dengan menggunakan is_prime

```python
def unique_int(input_list) :
    unique_list = []

    for num in input_list :
        if num not in unique_list :
            unique_list.append(num)
    
    return unique_list

# Use case Example
input_list = [4, 2, 4, 7, 9, 2, 1, 7]
result = unique_int(input_list)
print(result)
```

Explanation

unique_int bekerja dengan mengambil list yang dibarikan dan meng iterasi list tersebut dan dimasukan ke list baru, untuk angka yang sudah masuk tidak akan masuk lagi.

```python
rand_array = np.random.randint(1, 101, size=(5, 5))

print("Array awal:")
print(rand_array)

rand_array[rand_array % 2 == 0] = -1

print("\nArray setelah mengganti nilai genap:")
print(rand_array)
```

Explantion
