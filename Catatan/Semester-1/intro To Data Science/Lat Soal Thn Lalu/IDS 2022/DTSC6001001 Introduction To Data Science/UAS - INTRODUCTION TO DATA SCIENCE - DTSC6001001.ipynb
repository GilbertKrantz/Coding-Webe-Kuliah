```python
# type your codes here

raw_data_confirmed.keys()

confirm_data = raw_data_confirmed.drop(columns=['Province/State', 'Lat', 'Long'])

confirm_data = confirm_data.groupby("Country/Region").sum()

sum = confirm_data.sum(axis=1)

sum.name = 'Total'

total_data =  pd.DataFrame(sum).reset_index()

total_data

```
