# Import Tensorflow

```python
import tensorflow as tf
print(tf.__version__)
```

# Tensor Object

## Creating Tensor Object

```python
a = tf.constant([2,2])
a
```

```python
b = tf.constant([3,4])
b
```

## Variable vs Constant

### Variable

```python
a = tf.Variable([5,2])
a
```

```python
a.assign([1,2])
```

## Matric Calculation

```python
a = tf.constant([[1,2], [4, 2]])
b = tf.constant([[3,4], [5,6]])
```

### Addition and Subtraction

```python
c = tf.add(a,b)
c
```

```python
c = a+b
c
```

```python
d = tf.subtract(a,b)
d
```

```python
d = a-b
d
```

### Multiplication

#### Vector Multiplication

```python
e = tf.multiply(a,b)
e
```

#### Matrix Multiplication

```python
a @ b
```

```python
tf.matmul(a, b)
```

## Squeeze (Reduce Dimension)

```python
import numpy as np
```

```python
a = tf.constant(np.random.randint(0, 100, 224*224*3*2), shape=(2, 224, 224, 3))
print(a)
tf.squeeze(a)
```
