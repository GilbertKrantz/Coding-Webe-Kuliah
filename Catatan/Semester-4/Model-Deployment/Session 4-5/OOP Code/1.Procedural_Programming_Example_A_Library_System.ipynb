In a procedural style, you might have a data structure for storing book details and separate functions for each operation you want to perform on these books.

Data Structure for Books
You can use a dictionary to store book details:

```python
# Book data stored in a dictionary
books = {
    "123456789": {"title": "Harry Potter and the Philosopher's Stone",
                  "author": "J.K. Rowling", "checked_out": False}
    # More books can be added here
}
```

Functions for Operations
Then, you define separate functions to handle different actions:

```python
def check_out(isbn):
    if isbn in books and not books[isbn]['checked_out']:
        books[isbn]['checked_out'] = True
        print(f"{books[isbn]['title']} by {books[isbn]['author']} has been checked out.")
    else:
        print("Book not available for checkout.")

def return_book(isbn):
    if isbn in books and books[isbn]['checked_out']:
        books[isbn]['checked_out'] = False
        print(f"{books[isbn]['title']} by {books[isbn]['author']} has been returned.")
    else:
        print("This book was not checked out.")
```

Using the Functions
To perform operations, you call these functions with the appropriate ISBN:

```python
# Checking out a book
check_out("123456789")
# Output: "Harry Potter and the Philosopher's Stone by J.K. Rowling has been checked out."

# Returning the same book
return_book("123456789")
# Output: "Harry Potter and the Philosopher's Stone by J.K. Rowling has been returned."
```

In this procedural approach:

Data (book details) and functions (check_out, return_book) are separate.

Each function takes the necessary data (like the ISBN) as an argument.

The state of each book (whether it's checked out) is altered directly within the function.

This illustrates how procedural programming separates data and procedures, focusing on writing functions that operate on the data.

Unlike OOP, it doesn't encapsulate data and related operations within objects.
