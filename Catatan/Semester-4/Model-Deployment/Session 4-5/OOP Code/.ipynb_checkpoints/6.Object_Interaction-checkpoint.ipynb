```python
class BankAccount:
    # Class attribute
    interest_rate = 0.5  # A simple interest rate for all accounts

    def __init__(self, account_number, owner, balance=0):
        self._account_number = account_number  # Private attribute
        self.owner = owner
        self._balance = balance  # Private attribute

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            print("Balance cannot be negative.")
        else:
            self._balance = value

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
        else:
            self._balance += amount
            print(f"Deposited ${amount}. New balance: ${self._balance}.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self._balance:
            print("Insufficient funds.")
        else:
            self._balance -= amount
            print(f"Withdrew ${amount}. Remaining balance: ${self._balance}.")

    @classmethod
    def set_interest_rate(cls, rate):
        cls.interest_rate = rate
        print(f"New interest rate is: {cls.interest_rate}%")

    @staticmethod
    def validate_amount(amount):
        return amount > 0

    def apply_interest(self):
        if self.balance > 0 and BankAccount.validate_amount(self.balance):
            interest = self.balance * BankAccount.interest_rate / 100
            self.balance += interest
            print(f"Interest of ${interest} applied. New balance: ${self.balance}.")
```

```python
class Customer:
    def __init__(self, name):
        self.name = name
        self.accounts = []  # A list to store associated bank accounts

    def add_account(self, account):
        self.accounts.append(account)

    def get_account_info(self):
        for account in self.accounts:
            print(f"Account Number: {account.account_number}, Balance: ${account.balance}")
```

```python
# Creating a customer
alice = Customer("Alice")

# Creating bank accounts
account1 = BankAccount("12345", "Alice", 1000)
account2 = BankAccount("67890", "Alice", 500)

# Adding accounts to the customer
alice.add_account(account1)
alice.add_account(account2)

# Displaying account information for the customer
alice.get_account_info()
```
