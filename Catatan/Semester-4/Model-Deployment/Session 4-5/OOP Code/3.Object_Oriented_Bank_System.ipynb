To illustrate classes and objects, let's consider a bank account system.

Class (Blueprint): BankAccount - defines what a bank account is.
Objects (Instances): Each individual bank account is an object of the BankAccount class.

```python
class BankAccount:
    # Constructor method to initialize attributes
    def __init__(self, account_number, owner, balance=0):
        self.account_number = account_number
        self.owner = owner
        self.balance = balance

    # Method to deposit money
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ${amount}. New balance is ${self.balance}.")

    # Method to withdraw money
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        else:
            self.balance -= amount
            print(f"Withdrew ${amount}. Remaining balance is ${self.balance}.")
```

In this class:
__init__ is the constructor method. It initializes the object’s state with account_number, owner, and balance.
deposit and withdraw are methods that allow us to interact with the bank account's balance.


```python
# Creating two bank account objects
account1 = BankAccount("12345", "Alice")
account2 = BankAccount("67890", "Bob", 500)

# Interacting with these objects
account1.deposit(200)  # Deposits $200 into Alice's account
account1.withdraw(50)  # Withdraws $50 from Alice's account

account2.deposit(300)  # Deposits $300 into Bob's account
account2.withdraw(100)  # Withdraws $100 from Bob's account
```

Each BankAccount object (account1 and account2) has its own set of attributes (account_number, owner, balance) and can perform actions defined by the methods (deposit, withdraw).

Key Points:
Classes provide structure - they are a template for creating objects.
Objects are instances of classes; they hold actual data and can use the methods defined in the class.
Each object can have different values for its attributes, making them unique.
Methods inside a class allow objects to perform actions or operations on their data.
