Imagine a library system:

**Objects**: Each book in the library can be seen as an object.

**Class**: There's a generic blueprint for what a book is – this is the **class**. It defines properties like title, author, ISBN, etc.

**Attributes**: Specific details of each book, like "Harry Potter and the Philosopher's Stone", "J.K. Rowling", ISBN number, are attributes.

**Methods**: Operations such as 'Check out', 'Return', 'Reserve' a book, are methods.


```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn

    def check_out(self):
        print(f"{self.title} by {self.author} has been checked out.")

    def return_book(self):
        print(f"{self.title} by {self.author} has been returned.")
```

In this example:

Book is a class.
"__init__" is a special method called a constructor, used to initialize the book's attributes.
check_out and return_book are regular methods that define actions.

To use this class:

```python
# Creating an object of the class Book
harry_potter = Book("Harry Potter and the Philosopher's Stone", "J.K. Rowling", "123456789")

# Calling methods on this object
harry_potter.check_out()
# Output: "Harry Potter and the Philosopher's Stone by J.K. Rowling has been checked out."
harry_potter.return_book()
# Output: "Harry Potter and the Philosopher's Stone by J.K. Rowling has been returned."
```
