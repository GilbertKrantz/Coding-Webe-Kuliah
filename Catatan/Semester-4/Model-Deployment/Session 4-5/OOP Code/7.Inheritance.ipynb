Inheritance is a fundamental concept in object-oriented programming that allows a class to inherit attributes and methods from another class. It facilitates code reusability and creates a hierarchical relationship between classes.

Inheritance in the Bank System Example
To demonstrate inheritance, let's introduce two new types of bank accounts that extend the functionality of a basic BankAccount: a SavingsAccount and a CheckingAccount. Both these account types share common features with BankAccount but also have their unique characteristics.

Base Class: BankAccount
The BankAccount class will serve as our base class (or parent class).


```python
class BankAccount:
    def __init__(self, account_number, owner, balance=0):
        self.account_number = account_number
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New balance: ${self.balance}.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount}. Remaining balance: ${self.balance}.")
        else:
            print("Invalid withdrawal amount.")
```

Derived Classes: SavingsAccount and CheckingAccount
These classes will inherit from BankAccount and add specific features.

```python
class SavingsAccount(BankAccount):
    def __init__(self, account_number, owner, balance=0, interest_rate=0.5):
        super().__init__(account_number, owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate / 100
        self.balance += interest
        print(f"Interest of ${interest} added. New balance: ${self.balance}.")

class CheckingAccount(BankAccount):
    def __init__(self, account_number, owner, balance=0, overdraft_limit=100):
        super().__init__(account_number, owner, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        if amount <= self.balance + self.overdraft_limit:
            self.balance -= amount
            print(f"Withdrew ${amount}. Remaining balance: ${self.balance}.")
        else:
            print("Exceeded overdraft limit.")
```

Inheritance: SavingsAccount and CheckingAccount classes inherit from BankAccount. They get all the attributes and methods of BankAccount.
super().__init__: This function is used to call the constructor of the parent class. This way, the child class inherits the parent's initialisation process.
Extension: Both child classes extend the functionality. SavingsAccount adds interest-earning capability, and CheckingAccount adds overdraft protection.
Method Overriding: CheckingAccount overrides the withdraw method to include an overdraft feature, which changes the behavior of the method compared to the parent class.

Inheritance allows SavingsAccount and CheckingAccount to reuse the code from BankAccount, reducing redundancy and enhancing maintainability. Each child class preserves its unique characteristics while sharing common functionalities of the parent class.

```python
# Creating a Savings Account
savings = SavingsAccount("12345", "Alice", 1000)
savings.add_interest()  # Adds interest to the balance
savings.withdraw(200)  # Inherits withdrawal functionality from BankAccount

# Creating a Checking Account
checking = CheckingAccount("67890", "Bob", 500, 200)
checking.withdraw(600)  # Allowed due to overdraft limit
```
