```python
class BankAccount:
    # Class attribute
    interest_rate = 0.5  # A simple interest rate for all accounts

    def __init__(self, account_number, owner, balance=0):
        self.account_number = account_number
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ${amount}. New balance: ${self.balance}.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        else:
            self.balance -= amount
            print(f"Withdrew ${amount}. Remaining balance: ${self.balance}.")

    @classmethod
    def set_interest_rate(cls, rate):
        cls.interest_rate = rate
        print(f"New interest rate is: {cls.interest_rate}%")

    @staticmethod
    def validate_amount(amount):
        return amount > 0

    def apply_interest(self):
        if self.balance > 0 and BankAccount.validate_amount(self.balance):
            interest = self.balance * BankAccount.interest_rate / 100
            self.balance += interest
            print(f"Interest of ${interest} applied. New balance: ${self.balance}.")
```

```python
account1 = BankAccount("12345", "Alice", 1000)

# Using instance methods
account1.deposit(500)
account1.withdraw(200)
account1.apply_interest()  # Applies interest based on the current interest rate

# Using class method
BankAccount.set_interest_rate(1)  # Changes interest rate for all accounts

# Creating another account to see the effect of the new interest rate
account2 = BankAccount("67890", "Bob", 2000)
account2.apply_interest()  # Uses the updated interest rate
```
