# Import Module

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

# Data Visualization

## Mean, Median, Mode, Quantil, Percentile, Varians, Standart Deviation

```python
my_array = np.array([6, 7, 8])

print("Nilai Mahasiswa: ", my_array)
print("Jumlah Mahasiswa: ", len(my_array))
print("Mean Nilai Kuis: ", np.mean(my_array))
print("Median Nilai Kuis: ", np.median(my_array))
print("Modus Nilai Kuis: ", stats.mode(my_array))
print("Nilai Max: ", np.max(my_array))
print("Nilai Min: ", np.min(my_array))

# Quartal ada 2 cara, quartile dan percentile
print("Nilai Quartil(q1,q2,q2): ", np.quantile(my_array, [0.25, 0.5, 0.75]))
print("Nilai Quartil(q1,q2,q2): ", np.percentile(my_array, [25, 50, 75]))

print("Variance: ", np.var(my_array))
print("Standar Deviation: ", np.std(my_array))
```

## Contoh Latihan

```python
# Class A
array_1 = np.array([65, 68, 78, 54, 65, 
                    75, 56, 77, 83, 88, 
                    89, 73, 76, 86, 92, 
                    62, 75, 73, 78, 65, 
                    59, 61, 72, 65, 67, 
                    85, 63, 76, 77, 76, 
                    67, 89, 84, 86, 81, 
                    78, 79, 78])

# Class B
array_2 = np.array([85, 90, 90, 75, 90, 
                    90, 75, 80, 70, 70, 
                    80, 85, 80, 75, 80, 
                    70, 70, 80, 85, 80, 
                    75, 70, 70, 80, 85, 
                    59, 61, 72, 65, 67, 
                    85, 63, 76, 77, 76, 
                    90, 90, 60])


```

```python
print("A")
print("Nilai Mahasiswa: ", array_1)
print("Jumlah Mahasiswa: ", len(array_1))
print("Mean Nilai Kuis: ", np.mean(array_1))
print("Median Nilai Kuis: ", np.median(array_1))
print("Modus Nilai Kuis: ", stats.mode(array_1))
print("Nilai Quartil(q1,q2,q2): ", np.percentile(array_1, [25, 50, 75]))
print("Standar Deviation: ", np.std(array_1))

print()

print("B")
print("Nilai Mahasiswa: ", array_2)
print("Jumlah Mahasiswa: ", len(array_2))
print("Mean Nilai Kuis: ", np.mean(array_2))
print("Median Nilai Kuis: ", np.median(array_2))
print("Modus Nilai Kuis: ", stats.mode(array_2))
print("Nilai Quartil(q1,q2,q2): ", np.percentile(array_2, [25, 50, 75]))
print("Standar Deviation: ", np.std(array_2))
```

### BoxPlot

```python

plt.boxplot([array_1, array_2], notch = True, showmeans = True)
plt.title("Boxplot nilai uts Kelas A dan B")
plt.xlabel("Kelas")
plt.ylabel("Nilai")
plt.show()

```

### Histogram

```python


plt.hist(array_1, alpha = 0.5, range = (50, 95), bins = 9, label = "A")
plt.hist(array_2, alpha = 0.5, range = (50, 95), bins = 9, label = "B")
plt.title("Histogram nilai uts Kelas A dan B")
plt.xlabel("Nilai")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

### Genfromtext to get data from csv file

```python
#  genfromtxt()
# dataset is midterm result of class A, B, C, D
# What Insight Can you Get From This Data Set?
# What suggestion can you give to the teacher that teach?

dataset = np.genfromtxt("dataset_A.csv", delimiter = ",", dtype = str, encoding = "utf-8-sig")

A = dataset[:,0][1:].astype(int)
B = dataset[:,1][1:].astype(int)
C = dataset[:,2][1:].astype(int)
D = dataset[:,3][1:].astype(int)
```

### BoxPlot 

```python
# BOXPLIT A, B, C, D

plt.boxplot([A, B, C, D], notch = True, showmeans = True)
plt.title("Boxplot nilai uts Kelas A , B, C, D")
plt.xlabel("Kelas")
plt.ylabel("Nilai")
plt.show()
```

### Histogram


```python
# HISTOGRAM A, B, C, D

plt.hist(A, alpha = 0.5, range = (50, 100), bins = 10, label = "A")
plt.hist(B, alpha = 0.5, range = (50, 100), bins = 10, label = "B")
plt.hist(C, alpha = 0.5, range = (50, 100), bins = 10, label = "C")
plt.hist(D, alpha = 0.5, range = (50, 100), bins = 10, label = "D")
plt.title("Histogram nilai uts Kelas A , B, C, D")
plt.xlabel("Nilai")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
