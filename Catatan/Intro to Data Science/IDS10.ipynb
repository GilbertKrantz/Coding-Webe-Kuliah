```python
# Import Module

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Series
list_fruits = ['apple', 'banana', 'cherry', 'durian']

# Print the value of the 2nd index
print(list_fruits[2], '\n')

# Convert list to series
series_fruits = pd.Series(list_fruits)
print(series_fruits)
```

```python
# Series Indexing 1

# Changing the String index
series_capital = pd.Series(['Jakarta', 'Tokyo', 'London', 'France'], 
                           index=['Indonesia', 'Japan', 'UK', 'France'])

# Print the value of the series with the index UK
series_capital["UK"]
# Print the value of the series with the index 2
series_capital[2]
```

```python
# Series Indexing 2

# Changing the number index
series_animal = pd.Series(['Elang', 'Flamingo', 'Gajah', 'Ikan'],
                          index=[100, 100, 102, 103])

series_animal[102]
# series_animal[2]
series_animal[100]
```

```python
# Series Manipulation

# Reseting Index
# Use drop to reset to a series
# inplace to store the reset
series_animal.reset_index(drop=True, inplace=True)
series_animal 

series_gabungan = pd.concat([series_fruits, series_animal])

# Sort by Index
series_gabungan.sort_index(inplace=True)
series_gabungan.reset_index(inplace=True, drop=True)
series_gabungan

# Sort by Values
# series_gabungan.sort_values(inplace=True)
# series_gabungan.reset_index(inplace=True, drop=True)
# series_gabungan

series_gabungan[5] = 'Gorila'
series_gabungan[9] = 'Anjing'
series_gabungan['x'] = 'Brandon'

# Drop to remove value from a Series
series_gabungan.drop(labels=3, inplace=True)
series_gabungan
```
