```python
# Nomor 1
# Boleh gak usah bikin input
list1 = [1,2,3,4]
list2 = [4,5,6,7]

# Kalo Pake input
list3 = list(map(int, input().split()))
print(list3)

def periksa(list):
    count = 0
    for i in range(len(list)-1):
        if list[i] > list[i+1]:
            count -= 1
        elif list[i] < list[i+1]:
            count += 1
            
    if count == len(list)-1:
        return "Meningkat"
    elif count == (len(list)-1) * -1:
        return "Menurun"
    else:
        return "Tidak Keduanya"
    
print(periksa(list1))
print(periksa(list2))
print(periksa(list3))
```

```python
# Nomor 3

# A
a = np.arange(25)
print("Array: ",a)

# B
b = a.reshape(5,5)
print("Matrix: \n",b)

# C
c = b.sum(axis=1)
print("Sum: ",c)

# D
d = np.median(b, axis=0)
print("med: ",d)

# E
bigthanmean = []
for char in a:
    if char > a.mean():
        bigthanmean.append(char)
print("Bigger Than Mean: ",bigthanmean)
        
# F
genap = 0
ganjil = 0
for char in a:
    if char % 2 == 0:
        genap += 1
    else:
        ganjil += 1
print(genap, ganjil)
```

```python
# Sorting

print(series_siswa)
series_siswa2 = pd.Series(["Feli", "Alis", "Angel"])

series_gabungan = series_siswa.append(series_siswa2)

# inplace biar langsung masuk ke vaiable
# Drop biar Index awal tidak Kesimpan
# Sorting by Index
series_gabungan.sort_index(inplace=True)

# Reset Index Biar Indecnya gak double
# series_gabungan = series_gabungan.reset_index()
series_gabungan.reset_index(drop=True, inplace=True)

# Sort by Values
series_gabungan.sort_values(inplace=True)

series_gabungan.reset_index(drop=True, inplace=True)

print(series_gabungan)
print(series_gabungan[0])
```

```python
#Data Frame

dict_1 = {"Nama": "Ani", "Umur": 21, "WNI": True, "Gender": "Female", "GPA": 3.5}
dict_2 = {"Nama": "Budi", "Umur": 20, "WNI": True, "Gender": "Male", "GPA": 3.3}
dict_3 = {"Nama": "Charlie", "Umur": 23, "WNI": False, "Gender": "Male", "GPA": 3.2}
dict_4 = {"Nama": "Devi", "Umur": 20, "WNI": True, "Gender": "Female", "GPA": 3.7}

df_siswa = pd.DataFrame([dict_1, dict_2, dict_3, dict_4])
print(df_siswa)
```

```python
# Print Umur Siswa
print(df_siswa["Umur"])

# Print Siswa Dengan Lokasi 2
print(df_siswa.loc[2])

# Print Siswa Dengan Gender Male
print(df_siswa[df_siswa["Gender"] == "Male"])

# Print Mean dari umur siswa dengan GPA lebih besar sama dengan 3.5
print(df_siswa[df_siswa["GPA"] >= 3.5]["Umur"].mean())
```
